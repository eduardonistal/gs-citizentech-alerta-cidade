# Avaliação local do baseline, RAG e agente especializado

Execute no PowerShell, com Ollama ativo:

```powershell
$env:OLLAMA_MODEL='qwen2.5:7b'; node ../avaliacao/benchmark.mjs
```

Depois carregue `../avaliacao/results.json`. A rodada de 14/11/2026 com `qwen2.5:7b` está salva neste repositório.

````python
import json
from pathlib import Path
result = json.loads(Path('../avaliacao/results.json').read_text(encoding='utf-8'))
board = result['scoreboard']
print(f"model={result['model']}")
for mode in ('baseline', 'rag', 'agent'):
    print(f"{mode} task_compliance={board[mode]['task_compliance']:.3f}")
print(f"retrieval_recall_at_3={board['rag']['retrieval_recall_at_3']:.3f}")
result['scoreboard']
````
